```python
orders = [
    {"id": 2001, "customer": "Anan", "city": "Korat",
     "product": "keyboard", "qty": 2, "price": 450},
    {"id": 2002, "customer": "Benja", "city": "Bangkok",
     "product": "Mouse", "qty": 5, "price": 180},
    {"id": 2003, "customer": "Chai", "city": "Bangkok",
     "product": "Monitor", "qty": 1, "price": 3200},
    {"id": 2004, "customer": "Duang", "city": "Phuket",
     "product": "Mouse", "qty": 3, "price": 180},
    {"id": 2005, "customer": "Anan", "city": "Korat",
     "product": "Monitor", "qty": 2, "price": 3200},
    {"id": 2006, "customer": "Ekk", "city": "Bangkok",
     "product": "Keyboard", "qty": 4, "price": 450},
]

#สร้าง list สำหรับเก็บข้อมูลรายงาน
report = []
total_sales_before_vat = 0
total_vat = 0
total_sales = 0

for order in orders:
    total_price = order["qty"] * order["price"]
    vat = total_price * 0.07
    net_price = total_price + vat

    report.append((
        order["id"],
        round(total_price, 2),
        round(vat, 2),
        round(net_price, 2)
    ))

    # สะสมยอดรวมแต่ละส่วน
    total_sales_before_vat += total_price
    total_vat += vat
    total_sales += net_price

print("รายงานสรุปยอดขาย")
print("-" * 65)
print("รหัสคำสั่งซื้อ | ราคาก่อน VAT | VAT 7% | ราคาสุทธิ")
print("-" * 65)

for item in report:
    print(f"{item[0]:<15} {item[1]:>12,.2f} {item[2]:>12,.2f} {item[3]:>14,.2f}")

print("-" * 65)
# แสดงผลยอดสรุปรวมทั้งหมดแยกตามประเภท
print(f"ยอดขายรวมทั้งหมด (ก่อน VAT) = {total_sales_before_vat:,.2f} บาท")
print(f"ภาษีมูลค่าเพิ่มรวม (VAT 7%) = {total_vat:,.2f} บาท")
print(f"ยอดขายรวมทั้งหมด (รวม VAT)  = {total_sales:,.2f} บาท")
```
